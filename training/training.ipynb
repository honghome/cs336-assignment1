## Train TinyStories Transformer

Run the transformer training script on the tokenized TinyStories dataset.

```python
import subprocess
import sys
from pathlib import Path

# Get project root if this cell is run independently
project_root = Path.cwd().resolve()
for candidate in [project_root, *project_root.parents]:
    if (candidate / "pyproject.toml").exists() and (candidate / "cs336_basics").exists():
        project_root = candidate
        break

# Prefer the project virtualenv Python so numpy/torch are available.
venv_python = project_root / ".venv" / "Scripts" / "python.exe"
python_exe = str(venv_python) if venv_python.exists() else sys.executable

# Assignment hyperparameters from section 7.2.1.
vocab_size = 10_000
context_length = 256
d_model = 512
d_ff = 1344
theta = 10_000
num_layers = 4
num_heads = 16
batch_size = 32

# Target total tokens processed: batch_size * max_iters * context_length ~= 327,680,000.
total_tokens_target = 327_680_000
max_iters = total_tokens_target // (batch_size * context_length)

train_command = [
    python_exe,
    "-u",  # unbuffered stdout/stderr so notebook logs appear sooner
    str(project_root / "cs336_basics" / "transformer_train.py"),
    "--train_data", str(project_root / "training" / "TinyStoriesV2-GPT4-train_tokens.bin"),
    "--valid_data", str(project_root / "training" / "TinyStoriesV2-GPT4-valid_tokens.bin"),
    "--vocab_size", str(vocab_size),
    "--context_length", str(context_length),
    "--d_model", str(d_model),
    "--d_ff", str(d_ff),
    "--theta", str(theta),
    "--num_layers", str(num_layers),
    "--num_heads", str(num_heads),
    "--batch_size", str(batch_size),
    "--device", "cpu",
    "--max_iters", str(max_iters),
    "--log_every", "50",
    "--eval_every", "500",
    "--eval_batches", "20",
    "--ckpt_every", "1000",
    "--ckpt_path", str(project_root / "training" / "TinyStoriesV2-GPT4-transformer-checkpoint.pt"),
]

print(f"Total tokens processed target: {batch_size * max_iters * context_length:,}")
print(f"max_iters: {max_iters:,}")
print("Running:")
print(" ".join(train_command))
print()

process = subprocess.Popen(
    train_command,
    cwd=project_root,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

try:
    for line in process.stdout:
        print(line, end="")
finally:
    return_code = process.wait()

if return_code == 0:
    print("\nTraining command completed successfully.")
else:
    print(f"\nTraining command failed with exit code {return_code}.")
```
